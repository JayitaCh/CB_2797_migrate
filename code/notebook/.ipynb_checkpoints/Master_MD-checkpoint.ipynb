:::{important} **Objectives**
_To better understand how the spatial and temporal distribution of police response problems can be aligned to area level measures derived from health data_
:::

:::{tip} **Research Questions**
- [ ] 1. What are the spatial and temporal patterns of mental health related incidents recorded in West Yorkshire Police data across the Bradford region?
- [ ] 2. What are the spatial and temporal patterns of mental health interactions data across the Bradford region?
- [ ] 3. How do these patterns from police and health data overlap with, and depart from, each other? And how might we interpret these similarities / differences?
:::

:::{card} <font color= 'blue'> 1. Preparing the Police records for analysis </font>
:link: ./Merge_tables.ipynb
In this notebook, we join the datasets from BigQuery to transform it into a single dataset for the 2017-2022 timeframe. We also drop the duplicates and define the incident types previously categorised by _Dan and Amanda_. We finally validate our data based on the values observed in the Vulnerability Report. The wrangled dataset is then saved as a csv and can be accessed from [here](./BD_Policedata_2017_22_all.csv).
:::

:::{card}<font color= 'blue'> 2. Exploratory Analysis of the Police records</font>
:link: ./Police_mentalhealth_cleaneddata.ipynb
In this notebook, we check the daily, weekly, monthly and yearly trends of the police call records. We also join the dataset with the [LSOA polygons](./LSOA_2011_full.geojson) to create geo-spatial visualisations. Additionally, we also use the dataset created from the [following notebook](./Police_mentalhealth_LSOA_charac.ipynb) to generate a parallel coordinate plot and year-wise geo-sptial visualisation of the mental health related police calls.
:::

:::{card}<font color= 'blue'> 3. Preparing indicator list for LSOA clustering</font>
:link: ./Police_mentalhealth_LSOA_charac.ipynb
Since the dataset includes LSOA codes based on the 2011 Census, we harmonise these codes to align the 2021 population estimates with the 2011 LSOA boundaries. In this notebook, we also derive **nine** composite indicators for each LSOA to reduce dimensionality and extract meaningful characteristics related to mental health-related police calls. The [estimated indicators](./Estimated_indicators_MH_Police.csv) is then stored as a csv for conducting the cluster analysis subsequently.
:::

:::{card}<font color= 'blue'> 4. Cluster Analysis on Police records</font>
:link: ./Clustering_MH_police.ipynb
In this notebook, we examine the correlations among the estimated indicators and identify potential outliers. Since our focus is on understanding demand, we retain the outliers rather than removing them. We then apply two clustering algorithms, namely, K-means and K-medoids, both of which suggest three optimal clusters. However, the resulting clusters show limited quality in terms of magnitude and cardinality. Apart from the **percentage of non-crime incidents**, the other indicators contribute little to distinguishing meaningful cluster characteristics.
:::
